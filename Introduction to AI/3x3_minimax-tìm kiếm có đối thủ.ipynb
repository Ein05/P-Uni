```python
import math

# --- 1. Biểu diễn bàn cờ ---
# X = 1, O = -1, Trống = 0
def print_board(board):
    symbols = {1: 'X', -1: 'O', 0: ' '}
    print("\nBàn cờ hiện tại:")
    for row in board:
        print('|'.join(symbols[cell] for cell in row))
        print('-' * 5)
```

```python
# --- 2. Kiểm tra còn nước đi ---
def is_moves_left(board):
    for row in board:
        if 0 in row:
            return True
    return False
```

```python
# --- 3. Kiểm tra thắng/thua/hòa ---
def check_winner(board):
    # Kiểm tra hàng
    for row in board:
        if abs(sum(row)) == 3:
            return row[0]

    # Kiểm tra cột
    for col in range(3):
        column_sum = board[0][col] + board[1][col] + board[2][col]
        if abs(column_sum) == 3:
            return board[0][col]

    # Kiểm tra đường chéo
    diag1 = board[0][0] + board[1][1] + board[2][2]
    diag2 = board[0][2] + board[1][1] + board[2][0]
    if abs(diag1) == 3:
        return board[1][1]
    if abs(diag2) == 3:
        return board[1][1]

    # Nếu hết nước mà chưa có ai thắng thì hòa
    if not is_moves_left(board):
        return 0  # hòa

    return None  # chưa kết thúc
```

```python
# --- 4. Hàm Minimax ---
def minimax(board, depth, is_maximizing):
    result = check_winner(board)
    if result is not None:
        return result  # +1, -1 hoặc 0

    if is_maximizing:
        best = -math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == 0:
                    board[i][j] = 1  # X đi
                    value = minimax(board, depth + 1, False)
                    board[i][j] = 0
                    best = max(best, value)
        return best
    else:
        best = math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == 0:
                    board[i][j] = -1  # O đi
                    value = minimax(board, depth + 1, True)
                    board[i][j] = 0
                    best = min(best, value)
        return best
```

```python
# --- 5. Máy tính chọn nước đi tốt nhất ---
def find_best_move(board):
    best_val = math.inf
    best_move = (-1, -1)

    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                board[i][j] = -1  # máy (O) thử đi
                move_val = minimax(board, 0, True)
                board[i][j] = 0

                if move_val < best_val:
                    best_val = move_val
                    best_move = (i, j)

    return best_move
```

```python
# --- 6. Giao diện chơi ---
def main():
    board = [[0, 0, 0],
             [0, 0, 0],
             [0, 0, 0]]

    print("Trò chơi caro 3x3 (X vs O). Bạn là X.")
    print_board(board)

    while True:
        # Người chơi
        while True:
            try:
                row, col = map(int, input("Nhập vị trí (hàng cột, từ 0-2, cách nhau bởi dấu cách): ").split())
                if 0 <= row < 3 and 0 <= col < 3 and board[row][col] == 0:
                    board[row][col] = 1
                    break
                else:
                    print("⚠️ Nước đi không hợp lệ, thử lại!")
            except:
                print("⚠️ Nhập sai định dạng, thử lại!")

        print_board(board)
        if check_winner(board) is not None:
            break

        # Máy tính
        print("Máy tính đang suy nghĩ...")
        move = find_best_move(board)
        board[move[0]][move[1]] = -1
        print_board(board)

        if check_winner(board) is not None:
            break

    # Kết quả
    result = check_winner(board)
    if result == 1:
        print("🎉 Bạn thắng!")
    elif result == -1:
        print("💻 Máy thắng!")
    else:
        print("🤝 Hòa!")


if __name__ == "__main__":
    main()
```
